```python
# Sistema

import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.append(str(project_root / "src"))
```

```python
# Imports

import pandas as pd

from pangya_physics import CLUBS, Wind, find_velocity_for_distance
```

```python
# Validation

validation_path = project_root / "data" / "validation_cases.csv"

df_cases = pd.read_csv(validation_path)

df_cases
```

```python
# Resultados

results = []

for _, row in df_cases.iterrows():
    club = CLUBS[row["club"]]
    wind = Wind(
        speed=row["wind"],
        degree=row["degree"],
    )

    result = find_velocity_for_distance(
        target_distance=row["distance"],
        club=club,
        wind=wind,
    )

    results.append(
        {
            "case_id": row["case_id"],
            "club": row["club"],
            "distance": row["distance"],
            "wind": row["wind"],
            "degree": row["degree"],
            "python_velocity_z": result.velocity_z,
            "python_final_distance": result.final_distance,
            "python_error": result.error,
            "expected_power": row["expected_power"],
            "expected_desvio": row["expected_desvio"],
        }
    )

df_validation = pd.DataFrame(results)

df_validation
```
